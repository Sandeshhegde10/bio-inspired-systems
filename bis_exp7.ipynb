```python
import random


subjects = ["Maths", "Physics", "Chemistry", "English", "Computer"]
slots = ["Mon-9AM", "Mon-2PM", "Tue-9AM", "Tue-2PM", "Wed-9AM"]

POP_SIZE = 10
GENERATIONS = 50
MUTATION_RATE = 0.2



def fitness(timetable):
 
    conflicts = len(timetable) - len(set(timetable))
    return len(timetable) - conflicts



def create_timetable():
    return random.sample(slots, len(subjects))



def mutate(timetable):
    if random.random() < MUTATION_RATE:
        i, j = random.sample(range(len(timetable)), 2)
        timetable[i], timetable[j] = timetable[j], timetable[i]
    return timetable


def GEA():
    population = [create_timetable() for _ in range(POP_SIZE)]

    for generation in range(GENERATIONS):

        population.sort(key=fitness, reverse=True)
        best = population[0]

      
        new_population = [best]

        while len(new_population) < POP_SIZE:
            parent = random.choice(population[:5])
            child = parent.copy()

            child = mutate(child)
            new_population.append(child)

        population = new_population

    return population[0]



best_timetable = GEA()

print("Best Exam Timetable:")

for subject, slot in zip(subjects, best_timetable):
    print(subject, "->", slot)

print("Fitness:", fitness(best_timetable))
```
